```python
import numpy as np
import matplotlib.pyplot as plt

# Set the number of data points
n = 1000

# Set a seed for reproducibility (ensures the same random numbers are generated every time)
np.random.seed(42)

# Generate training data: 1000 samples with 15 features, drawn from a normal distribution (mean=2, std=8)
x_train = np.random.normal(2, 8, size=(n, 15))

# Define the true weight vector (coefficients for the linear function)
w_true = np.array([35, 67, 4, 5, 67, 81, -4, 98, 12, 23, 54, 48, 23, -65, 12])

# Compute the corresponding y values for the training set using the dot product
y_train = np.dot(x_train, w_true)

# Generate validation data (new set of x values with the same distribution as training data)
x_validate = np.random.normal(2, 8, size=(n, 15))

# Compute the corresponding y values for the validation set using the true weights
y_validate = np.dot(x_validate, w_true)

# Compute the weight vector using the Normal Equation: w = (X^T * X)^(-1) * X^T * y
w_estimated = np.linalg.inv(x_train.T @ x_train) @ x_train.T @ y_train

# Compute predicted y values for the validation dataset
y_pred = np.dot(x_validate, w_estimated)

# Compute Sum of Squared Errors (SSE) for validation data
sse = sum((y_validate - y_pred) ** 2)

# Plot the actual vs predicted y values for the validation set
plt.figure(figsize=(8, 6))
plt.scatter(y_validate, y_pred, alpha=0.5, label='Predictions')  # Scatter plot of actual vs predicted values
plt.plot([min(y_validate), max(y_validate)], [min(y_validate), max(y_validate)], color='red', linestyle='--', label='Ideal Fit')  # Ideal fit line (y = x)
plt.xlabel('Actual Y')
plt.ylabel('Predicted Y')
plt.title('Actual vs Predicted Values')
plt.legend()
plt.grid(True)
plt.show()

# Print the estimated equation of the fitted line
equation = "y = " + " + ".join([f"{w:.2f}*x{i}" for i, w in enumerate(w_estimated)])
print("Estimated Equation of the Line:")
print(equation)

# Print the Sum of Squared Errors (SSE)
print(f"Sum of Squared Errors (SSE): {sse}")
```
